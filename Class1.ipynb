```python
# %s - STR
# %d - INT
# %f - FLOAT
```

```python
# Method 2 - % Method
name = "ammar"
age = 30
#print("My name is ",name , "and my age is ",age )
print("my name is %s and my age is %d" %(name,age)) # optimal method
```

```python
# Method 2 - % Method
name = "ammar"
income = 53678.89
print("my name is %s and my income is %f" %(name,income))
```

```python
# Method 2 - % Method
name = "ammar"
income = 53678.89
print("my name is %s and my income is %d" %(name,income)) #
```

```python
# Method 2 - % Method
name = "ammar"
income = 53678.89
print("my name is %s and my income is %.2f" %(name,income)) # control the decimal
```

```python
# Method 3 - .format method
name = "ammar"
age = 30
#print("my name is {} and my age is {}".format(name,age))
print("my name is {0} and my age is {1}".format(name,age)) # index it for more flexibility
print("my name is {x} and my age is {y}".format(x = name, y = age))
print("my name is {} and my age is {:.2f}".format(name, age))
```

```python
# Method 4 - f-string (most common)
name = "ammar"
age = 30
print(f"my name is {name} and my age is {age}")
```

```python
# Method 4 - f-string (most common)
name = "ammar"
income = 33432.89
print(f"my name is {name} and my age is {income: .2f}")
```

## Formatting functions

```python
name = "  Ugandhar Vaddi  "

print(name.strip())           # Remove outer whitespace
print(name.upper())           # UGAN... in uppercase
print(name.lower())           # ugandhar vaddi
print(name.replace("Vaddi", "Kumar"))
print(name.count("a"))        # Count occurrences
print(name.startswith("  U")) # True
```

## List functions

image.png

```python
a = [10, 20, 30]
b = a

b.append(40)

print(a)
# [10, 20, 30, 40]
```

```python
a = [10, 20, 30]
b = a.copy()

b.append(40)

print(a)  # [10, 20, 30]
print(b)  # [10, 20, 30, 40]
```

```python
#[expression for item in iterable if condition]

marks = [85, 92, 76, 92, 65]

high_marks = [mark for mark in marks if mark >= 80]

print(high_marks)
# [85, 92, 92]
```

```python
numbers = [1, 2, 3, 4, 5]

squares = [n ** 2 for n in numbers]

print(squares)
# [1, 4, 9, 16, 25]
```

```python
# set() - produces only unique values

values = [12, 25, 8, 25, 30]

unique_values = sorted(set(values))

print(unique_values)
print(len(unique_values))
print(f"Average: {sum(values) / len(values):.2f}")
```
